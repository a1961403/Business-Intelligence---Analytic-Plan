# 2. Data Analysis

2.1. Identify most mentioned skills

```python
import matplotlib.pyplot as plt

# Top 15 most frequent requirements (unstandardised)
top_15_skills = (skill_df["job_skills"].value_counts().head(15))

plt.figure(figsize=(10, 6))

top_15_skills.sort_values().plot(kind="barh",color="steelblue")

plt.title("Top 15 Requirements in Business Analyst Job Advertisements")
plt.xlabel("Frequency")
plt.ylabel("Skill / Requirement")

plt.tight_layout()
plt.show()
```

```python
top_15_skills 
```

```python
skill_df_chart = skill_df.copy()

# Merge obvious duplicates
skill_df_chart.loc[skill_df_chart["job_skills"].str.contains("communication", na=False),"job_skills"] = "communication"

skill_df_chart.loc[skill_df_chart["job_skills"].str.contains("problem", na=False),"job_skills"] = "problem solving"
```

```python
top_15_skills = skill_df_chart["job_skills"].value_counts().head(15)

top_15_skills
```

```python
# check the strings that contain "communication"
skill_df[skill_df["job_skills"].str.contains("communication",na=False)]["job_skills"].value_counts()
```

```python
# check the strings that contain "problem" for "problem solving"
skill_df[skill_df_chart["job_skills"].str.contains("problem", na=False)]["job_skills"].value_counts()
```

```python
skill_df_chart = skill_df.copy()

skill_df_chart["job_skills"] = skill_df_chart["job_skills"].replace({
    "communication skills": "communication",
    "problemsolving": "problem solving"
})
```

```python
# the next skills added to the list are "jira" and "collaboration"

top_15_2 = skill_df_chart["job_skills"].value_counts().head(15)

top_15_2
```

```python
plt.figure(figsize=(10,6))


ax = top_15_2.sort_values().plot(kind="barh",color="steelblue")

plt.title("Top 15 Requirements in Business Analyst Job Advertisements")
plt.xlabel("Frequency")
plt.ylabel("Skill / Requirement")

# Display frequency values beside each bar
for i, value in enumerate(top_15_2.sort_values()):
    ax.text(value + 1,i,str(value),va="center")

plt.tight_layout()

plt.savefig("top_15_skills.png")

plt.show()
```

2.2. Identify employers that frequently hire Business Analysts

```python
top_employers = ba_df["company"].value_counts().head(15)

plt.figure(figsize=(10, 6))

ax = top_employers.sort_values().plot(kind="barh",color="steelblue")

plt.title("Top Employers Hiring Business Analysts")
plt.xlabel("Number of Job Advertisements")
plt.ylabel("Employer")

# Display frequency values beside each bar
for i, value in enumerate(top_employers.sort_values()):
    ax.text(value + 1,i,str(value),va="center")

plt.tight_layout()

plt.savefig("top_15_employers.png")

plt.show()
```

2.3. Identify locations that had the most BA job advertisements

```python
top_locations = ba_df["job_location"].value_counts().head(15)

top_locations
```

```python
plt.figure(figsize=(10, 6))

ax = top_locations.sort_values().plot(kind="barh",color="steelblue")

plt.title("Most Frequently Recorded Job Locations")
plt.xlabel("Number of Job Advertisements")
plt.ylabel("Location")

for i, value in enumerate(top_locations.sort_values()):
    ax.text(value + 1, i, str(value), va="center")

plt.tight_layout()

plt.savefig("top_15_location.png")

plt.show()
```
